```python
import  random
```

```python
# 1  ile  100 arasında  rastgele  bir sayı seçelim  
```

```python
gizli_sayi = random.randint(1,100)
```

```python
deneme_sayisi = 0
```

```python
print("1 ile 100 arasında  bir sayi tuttum.Bakalim bulabilcek misin?")
```

```python

while True:
    tahmin = int(input("Tahminin nedir?: "))
    deneme_sayisi += 1

    if tahmin < gizli_sayi:
        print("Daha yüksek bir sayı söyle!")
    elif tahmin > gizli_sayi:
        print("Daha düşük bir sayı söyle!")
    else:
        print(f"Tebrikler! {gizli_sayi} sayısını {deneme_sayisi} denemede buldun.")
        break
        
```
